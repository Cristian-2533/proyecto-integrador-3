Este notebook carga el dataset `scikit-learn/iris` desde Hugging Face, transforma sus datos y genera un resumen `.txt` del DataFrame final.

Flujo de trabajo:
1. Importar las clases reutilizables del proyecto.

Ejecuta las celdas en orden. El dataset es p?blico y normalmente no requiere autenticaci?n.

## 1. Preparar el entorno e importar las clases

El notebook busca la ra?z del proyecto para importar `HFDatasetLoader` y `Reportes`. Tambi?n define rutas estables para guardar los archivos aunque el notebook se ejecute desde otra carpeta.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src" / "clases_po").exists():
    ROOT = ROOT.parent

if not (ROOT / "src" / "clases_po").exists():
    raise FileNotFoundError("No se encontr? src/clases_po desde el directorio actual.")

sys.path.insert(0, str(ROOT / "src"))

from clases_po import HFDatasetLoader, Reportes

DATA_DIR = ROOT / "src" / "sincronicas" / "datos"
REPORT_DIR = ROOT / "src" / "sincronicas" / "reportes"
```

## 2. Configurar y cargar el dataset

`scikit-learn/iris` contiene mediciones de s?palos y p?talos para tres especies de iris. `HFDatasetLoader` obtiene la divisi?n `train`, la convierte a DataFrame y guarda una copia CSV de los datos originales.

```python
loader = HFDatasetLoader(
    dataset_name="scikit-learn/iris",
    split="train",
    filename=DATA_DIR / "iris.csv",
)

df = loader.load()
print(f"Dataset cargado: {len(df)} filas y {len(df.columns)} columnas")
df.head()
```

## 3. Revisar la estructura original

Antes de transformar, revisamos las columnas, sus tipos y los valores no nulos del DataFrame descargado.

```python
print(f"Filas: {len(df)}")
print(f"Columnas: {len(df.columns)}")
df.info()
```

## 4. Transformaci?n 1: retirar el identificador

La columna `Id` solo identifica cada fila y no es una medici?n de la flor, por eso se elimina.

```python
df_transformado = df.drop(columns=["Id"])
df_transformado.head()
```

## 5. Transformaci?n 2: renombrar columnas

Renombramos las variables al espa?ol para que el DataFrame sea m?s f?cil de interpretar.

```python
df_transformado = df_transformado.rename(columns={
    "SepalLengthCm": "longitud_sepalo_cm",
    "SepalWidthCm": "ancho_sepalo_cm",
    "PetalLengthCm": "longitud_petalo_cm",
    "PetalWidthCm": "ancho_petalo_cm",
    "Species": "especie",
})
df_transformado.head()
```

## 6. Transformaci?n 3: filtrar una especie

Seleccionamos las flores `Iris-setosa` para observar un subconjunto espec?fico. El resultado contiene 50 filas.

```python
df_transformado = df_transformado.loc[
    df_transformado["especie"] == "Iris-setosa"
].copy()
print(f"Filas despu?s del filtro: {len(df_transformado)}")
df_transformado.head()
```

## 7. Transformaci?n 4: crear una variable derivada

Calculamos un ?rea aproximada del p?talo multiplicando su longitud por su ancho, ambas medidas en cent?metros.

```python
df_transformado["area_petalo_aprox_cm2"] = (
    df_transformado["longitud_petalo_cm"] * df_transformado["ancho_petalo_cm"]
)
df_transformado.head()
```

## 8. Generar el resumen del DataFrame transformado

La clase `Reportes` recibe el DataFrame final, muestra sus dimensiones, tipos de columnas, valores nulos y estad?sticas num?ricas.

```python
reportes = Reportes()
texto = reportes.generar_reporte(df_transformado)
print(texto)
```

## 9. Guardar el resumen en un archivo `.txt`

El reporte corresponde al DataFrame transformado (filtrado a `Iris-setosa` y con la variable derivada) y queda guardado en `src/sincronicas/reportes/iris.txt`.

```python
ruta = reportes.guardar_reporte(
    df_transformado,
    REPORT_DIR / "iris.txt",
)
print(f"Reporte guardado en: {ruta}")
```
